## Plotting turbidity and constituent relationships across La Jara Samples fro Supporting Information
Note: Hex codes for viridis colors: 
- #440154 (Dark Purple / Violet)
- #3b528b (Deep Blue)
- #21918c (Teal)
- #5ec962 (Light Green)
- #fde725 (Yellow)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import linregress
from matplotlib.lines import Line2D
from matplotlib import colormaps
viridis = colormaps["viridis"]

df = pd.read_csv("constituent_data.csv")
# get rid of accidental spaces from column names
df.columns = df.columns.str.strip()
numeric_cols = ["SS", "POC", "DOC", "PP", "SRP", "Turbidity", "fDOM"]
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")
df["Sample ID"] = df["Sample ID"].astype(str).str.strip().str.lower()

# assign each observation to a sample group
def assign_group(sample_id):
    if sample_id.startswith("storm"):
        return "Summer storms"
    elif sample_id == "spring" or sample_id.startswith("spring"):
        return "Spring events"
    elif sample_id.startswith("experiment"):
        return "Experiments"

df["Group"] = df["Sample ID"].apply(assign_group)
print(df["Group"].value_counts(dropna=False))
```

Functions

```python
def calculate_r2(data, x_col, y_col):
    paired = data[[x_col, y_col]].dropna()
    if len(paired) < 2:
        return np.nan
    if paired[x_col].nunique() < 2 or paired[y_col].nunique() < 2:
        return np.nan
    regression = linregress(paired[x_col], paired[y_col])
    return regression.rvalue ** 2

def format_r2(value):
    if pd.isna(value):
        return "NA"
    return f"{value:.2f}"

def plot_relationship(ax, data, x_col, y_col, x_label, limx, r2_xy):
    # colors consistent across all four panels
    group_colors = {
        "Summer storms": viridis(0.7),  # green
        "Spring events": viridis(0.35),  # teal
        "Experiments":   viridis(0.90),  # yellow
    }
    group_order = ["Experiments", "Summer storms", "Spring events", ]
    all_data_color = viridis(0.1)       # dark purple
    group_markers = {
        "Summer storms": "o",
        "Spring events": "s",
        "Experiments": "^",
    }

    panel_data = data[[x_col, y_col, "Group"]].dropna().copy()
    # scatter plots observations by group
    for group in group_order:
        subset = panel_data[panel_data["Group"] == group]
        ax.scatter(subset[x_col], subset[y_col], s=42, color=group_colors[group], edgecolor="black", marker=group_markers[group],
            linewidth=0.4, alpha=1, zorder=3, label=group)
    # full-dataset regression line only
    if (len(panel_data) >= 2
        and panel_data[x_col].nunique() >= 2
        and panel_data[y_col].nunique() >= 2
    ):
        fit = linregress(panel_data[x_col], panel_data[y_col])
        x_line = np.linspace(panel_data[x_col].min(), panel_data[x_col].max(), 200)
        y_line = fit.intercept + fit.slope * x_line
        ax.plot(x_line, y_line, linestyle="--", linewidth=1.8, zorder=2, color="#440154")

    # calculate the four R² values
    r2_all = calculate_r2(panel_data, x_col, y_col)

    r2_summer = calculate_r2(panel_data[panel_data["Group"] == "Summer storms"], x_col, y_col)
    r2_spring = calculate_r2(panel_data[panel_data["Group"] == "Spring events"], x_col, y_col)
    r2_experiments = calculate_r2(panel_data[panel_data["Group"] == "Experiments"], x_col, y_col)
    r2_entries = [
        ("All", r2_all, all_data_color),
        ("Summer", r2_summer, group_colors["Summer storms"]),
        ("Spring", r2_spring, group_colors["Spring events"]),
        ("Experiments", r2_experiments, group_colors["Experiments"]),
    ]
    highlight_colors = {
        "All": viridis(0.1),          # light purple
        "Summer": viridis(0.7),       # light green
        "Spring": viridis(0.35),       # light teal
        "Experiments": viridis(0.9)   # light yellow
    }

    entries = [
        ("All", r2_all),
        ("Summer", r2_summer),
        ("Spring", r2_spring),
        ("Experiments", r2_experiments),
    ]

    x_text, y_start = r2_xy
    dy = 0.08 # vertical distance between R² entries

    for label, value in entries:
        ax.text(
            x_text,
            y_start,
            rf"{label}: $R^2={format_r2(value)}$",
            transform=ax.transAxes,
            ha="right",
            va="top",
            fontsize=9,
            color="black",
            bbox=dict(
                facecolor=highlight_colors[label],
                edgecolor="none",
                boxstyle="round,pad=0",
                alpha=0.30
            ),
            zorder=10
        )
        y_start -= dy

    # panel formatting
    ax.set_xlabel(x_label)
    ax.set_xlim(-5, limx)
    ax.set_ylabel(f"{y_col} (mg/L)")

    ax.grid(True, alpha=0.25)
    ax.set_axisbelow(True)
```

```python
def plot_relationship(ax, data, x_col, y_col, x_label, limx, r2_xy):
    # colors consistent across panels
    group_colors = {
        "Summer storms": viridis(0.7),
        "Spring events": viridis(0.35),
        "Experiments":   viridis(0.90),
    }
    group_order = ["Experiments", "Summer storms", "Spring events"]
    all_data_color = viridis(0.1)
    group_markers = {
        "Summer storms": "o",
        "Spring events": "s",
        "Experiments": "^",
    }

    panel_data = data[[x_col, y_col, "Group"]].dropna().copy()

    # prepare twin top x-axis only for Turbidity
    ax_top = None
    if x_col == "Turbidity":
        spring_vals = data[data["Group"] == "Spring events"][x_col].dropna()
        ax.set_xlim(-5, limx)
        if len(spring_vals) >= 1:
            tpad = 0.05 * (spring_vals.max() - spring_vals.min() if spring_vals.max() != spring_vals.min() else spring_vals.max() or 1)
            xmin_ntu = spring_vals.min() - tpad
            xmax_ntu = spring_vals.max() + tpad
            ax_top = ax.twiny()
            ax_top.set_xlim(xmin_ntu, xmax_ntu)
            ax_top.set_xlabel("Turbidity (NTU)")
            ax_top.grid(False)
        else:
            # no spring values, keep only bottom axis
            ax_top = None
    else:
        ax.set_xlim(-5, limx)

    # scatter: plot Spring events on top axis (NTU) if available, others on bottom (FNU)
    for group in group_order:
        subset = panel_data[panel_data["Group"] == group]
        if subset.empty:
            continue
        plot_ax = ax_top if (x_col == "Turbidity" and group == "Spring events" and ax_top is not None) else ax
        plot_ax.scatter(
            subset[x_col],
            subset[y_col],
            s=42,
            color=group_colors[group],
            edgecolor="black",
            marker=group_markers[group],
            linewidth=0.4,
            alpha=1,
            zorder=3,
            label=group
        )

    # full-dataset regression line (keeps previous behavior)
    if (len(panel_data) >= 2
        and panel_data[x_col].nunique() >= 2
        and panel_data[y_col].nunique() >= 2
    ):
        fit = linregress(panel_data[x_col], panel_data[y_col])
        x_line = np.linspace(panel_data[x_col].min(), panel_data[x_col].max(), 200)
        y_line = fit.intercept + fit.slope * x_line
        ax.plot(x_line, y_line, linestyle="--", linewidth=1.8, zorder=2, color=all_data_color)

    # calculate the four R² values (same logic as before)
    r2_all = calculate_r2(panel_data, x_col, y_col)
    r2_summer = calculate_r2(panel_data[panel_data["Group"] == "Summer storms"], x_col, y_col)
    r2_spring = calculate_r2(panel_data[panel_data["Group"] == "Spring events"], x_col, y_col)
    r2_experiments = calculate_r2(panel_data[panel_data["Group"] == "Experiments"], x_col, y_col)

    highlight_colors = {
        "All": viridis(0.1),
        "Summer": viridis(0.7),
        "Spring": viridis(0.35),
        "Experiments": viridis(0.9)
    }

    entries = [
        ("All", r2_all),
        ("Summer", r2_summer),
        ("Spring", r2_spring),
        ("Experiments", r2_experiments),
    ]

    x_text, y_start = r2_xy
    dy = 0.08

    for label, value in entries:
        ax.text(
            x_text,
            y_start,
            rf"{label}: $R^2={format_r2(value)}$",
            transform=ax.transAxes,
            ha="right",
            va="top",
            fontsize=9,
            color="black",
            bbox=dict(
                facecolor=highlight_colors[label],
                edgecolor="none",
                boxstyle="round,pad=0",
                alpha=0.30
            ),
            zorder=10
        )
        y_start -= dy

    # axis formatting and disable gridlines
    ax.set_xlabel(x_label)
    ax.set_ylabel(f"{y_col} (mg/L)")
    ax.grid(False)
    if ax_top is not None:
        ax_top.grid(False)
    ax.set_axisbelow(True)
```

```python
sns.set_style("whitegrid")

fig, axes = plt.subplots(1, 2, figsize=(8, 4))

plot_relationship(
    ax=axes[0],
    data=df,
    x_col="Turbidity",
    y_col="SS",
    x_label="Turbidity (FNU)",
    limx=df["Turbidity"].max() * 1.05,
    r2_xy=(0.97, 0.45)
)

plot_relationship(
    ax=axes[1],
    data=df,
    x_col="fDOM",
    y_col="DOC",
    x_label="fDOM (RFU)",
    limx=df["fDOM"].max() * 1.05,
    r2_xy=(0.97, 0.45)   # or (0.97, 0.97) if you prefer
)
fig.savefig("figure3.png", dpi=300, bbox_inches="tight")
```
