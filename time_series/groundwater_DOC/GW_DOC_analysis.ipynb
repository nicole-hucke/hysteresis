```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
import seaborn as sns
from matplotlib.ticker import StrMethodFormatter
import matplotlib.dates as mdates

directory = '../../data/groundwater'
WSE_dict = {}

for filename in os.listdir(directory): # iterate over all files in GW_and_Piezos folder
    if filename.endswith('.csv'): # checking if its a csv
        # construct the full file path
        file_path = os.path.join(directory, filename)
        df = pd.read_csv(file_path)
        df['Date_Time'] = pd.to_datetime(df['Date_Time'])
        # remove the '.csv' from the filename to use as the dictionary key
        key = filename[:-4]
        # store the data frame in the dictionary
        WSE_dict[key] = df
```

### Calculate hydraulic head (groundwater WSE - piezometer WSE)

```python
# For each piezometer, calculate VHG in each year
# VHG = groundwater WSE - piezometer WSE

vhg_results = {}
year_config = {
    2023: {
        "down": {
            "gw_left": "GW_down_left_2023",
            "gw_right": "GW_down_right_2023",
            "piezos": ["P1", "P2", "P3"],
            "piezo_suffix": "2023",
        },
    },
    2022: {
        "down": {
            "gw_left": "GW_down_left_2022",
            "gw_right": "GW_down_right_2022",
            "piezos": ["P1", "P2", "P3"],
            "piezo_suffix": "2022",
        },
    },
    2021: {
        "down": {
            "gw_left": "GW_down_left_2021",
            "gw_right": "GW_down_right_2021",
            "piezos": ["P1", "P2", "P3"],
            "piezo_suffix": "2021",
        },
    },
}
for year, banks in year_config.items():
    for bank, cfg in banks.items():
        gw_left = WSE_dict[cfg["gw_left"]][["Date_Time"] + cfg["piezos"]].rename(
            columns={p: f"{p}_gw_left" for p in cfg["piezos"]}
        )
        gw_right = WSE_dict[cfg["gw_right"]][["Date_Time"] + cfg["piezos"]].rename(
            columns={p: f"{p}_gw_right" for p in cfg["piezos"]}
        )

        for piezo in cfg["piezos"]:
            piezo_key = f"{piezo}C_{cfg['piezo_suffix'][-2:]}_WSE"
            piezo_df = WSE_dict[piezo_key][["Date_Time", "WSE"]].rename(columns={"WSE": f"{piezo}_WSE"})

            merged = (
                piezo_df.merge(gw_left[["Date_Time", f"{piezo}_gw_left"]], on="Date_Time", how="inner")
                        .merge(gw_right[["Date_Time", f"{piezo}_gw_right"]], on="Date_Time", how="inner")
            )

            merged[f"{piezo}_h_left"] = merged[f"{piezo}_gw_left"] - merged[f"{piezo}_WSE"]
            merged[f"{piezo}_h_right"] = merged[f"{piezo}_gw_right"] - merged[f"{piezo}_WSE"]
            merged[f"{piezo}_h_average"] = merged[[f"{piezo}_h_left", f"{piezo}_h_right"]].mean(axis=1)

            vhg_results[f"{piezo}_{bank}_{year}"] = merged[[
                "Date_Time",
                f"{piezo}_WSE",
                f"{piezo}_gw_left",
                f"{piezo}_gw_right",
                f"{piezo}_h_left",
                f"{piezo}_h_right",
                f"{piezo}_h_average"
            ]]

vhg_results["P1_down_2023"].head()
```

```python
# export data as csv files
output_dir = 'head_calculations'
for key, df in vhg_results.items():
    output_path = os.path.join(output_dir, f"{key}_hydraulic_head.csv")
    df.to_csv(output_path, index=False)
```

Calculating seasonal averages

```python
# only keep VHF averages for each piezometer: 
# for 2023
full_2023 = pd.DataFrame()
for key, df in vhg_results.items():
    if "2023" in key:
        piezo = key.split("_")[0]
        bank = key.split("_")[1]
        df_subset = df[["Date_Time", f"{piezo}_h_left", f"{piezo}_h_right", f"{piezo}_h_average"]]
        if full_2023.empty:
            full_2023 = df_subset
        else:
            full_2023 = full_2023.merge(df_subset, on="Date_Time", how="outer")

# for 2022
full_2022 = pd.DataFrame()
for key, df in vhg_results.items():
    if "2022" in key:
        piezo = key.split("_")[0]
        bank = key.split("_")[1]
        df_subset = df[["Date_Time", f"{piezo}_h_left", f"{piezo}_h_right", f"{piezo}_h_average"]]
        if full_2022.empty:
            full_2022 = df_subset
        else:
            full_2022 = full_2022.merge(df_subset, on="Date_Time", how="outer")

# for 2021
full_2021 = pd.DataFrame()
for key, df in vhg_results.items():
    if "2021" in key:
        piezo = key.split("_")[0]
        bank = key.split("_")[1]
        df_subset = df[["Date_Time", f"{piezo}_h_left", f"{piezo}_h_right", f"{piezo}_h_average"]]
        if full_2021.empty:
            full_2021 = df_subset
        else:
            full_2021 = full_2021.merge(df_subset, on="Date_Time", how="outer")

full_2021
```

```python
# separate 2023 data into seasons to calculate averages
spring_2023 = full_2023[(full_2023["Date_Time"] >= "2023-03-01") & (full_2023["Date_Time"] < "2023-06-19")]
summer_2023 = full_2023[(full_2023["Date_Time"] >= "2023-06-25") & (full_2023["Date_Time"] < "2023-09-21")]
# separate 2022 and 2021 to only have summer data for comparison
summer_2022 = full_2022[(full_2022["Date_Time"] >= "2022-06-22") & (full_2022["Date_Time"] < "2022-09-22")]
summer_2021 = full_2021[(full_2021["Date_Time"] >= "2021-06-21") & (full_2021["Date_Time"] < "2021-09-21")]

# export as csv files
spring_2023.to_csv(os.path.join(output_dir, "spring_2023_hydraulic_head.csv"), index=False)
summer_2023.to_csv(os.path.join(output_dir, "summer_2023_hydraulic_head.csv"), index=False)
summer_2022.to_csv(os.path.join(output_dir, "summer_2022_hydraulic_head.csv"), index=False)
summer_2021.to_csv(os.path.join(output_dir, "summer_2021_hydraulic_head.csv"), index=False)
```

# Spring DOC vs GW 

```python
# sonde data
down_sonde = pd.read_csv('../../data/sonde_data/sonde_down_full_record_smoothed.csv', parse_dates=['DateTime'], index_col='DateTime')
up_sonde = pd.read_csv('../../data/sonde_data/sonde_up_full_record_smoothed.csv', parse_dates=['DateTime'], index_col='DateTime')
# nutrient data
down_constituents = pd.read_csv('../../data/constituents/SP23_down_constituents.csv', parse_dates=['Date_Time'], index_col='Date_Time')
up_constituents = pd.read_csv('../../data/constituents/SP23_up_constituents.csv', parse_dates=['Date_Time'], index_col='Date_Time')

# determine dates of spring: 
min_date = pd.Timestamp('2023-04-18')
max_date = pd.Timestamp('2023-05-05')
down_sonde = down_sonde.loc[(down_sonde.index >= min_date) & (down_sonde.index <= max_date)]
up_sonde = up_sonde.loc[(up_sonde.index >= min_date) & (up_sonde.index <= max_date)]

# in sonde data sets, delete rows with negative data 
down_sonde = down_sonde[(down_sonde['Turbidity FNU'] >= 0) & (down_sonde['fDOM RFU'] >= 0) & (down_sonde['fDOM RFU'] <= 2)]
up_sonde = up_sonde[(up_sonde['Turbidity FNU'] >= 0) & (up_sonde['fDOM RFU'] >= 0) & (up_sonde['fDOM RFU'] <= 3)]

# determine dates of spring for GW data:
min_date = pd.Timestamp('2023-04-01')
max_date = pd.Timestamp('2023-05-15')
spring_2023 = spring_2023[(spring_2023["Date_Time"] >= min_date) & (spring_2023["Date_Time"] <= max_date)]
doc_col = "DOC (mg/L)"
fdom_col = "fDOM RFU"

# sonde resampling
down_sonde = down_sonde[~down_sonde.index.duplicated(keep="first")]
up_sonde = up_sonde[~up_sonde.index.duplicated(keep="first")]
down_sonde = down_sonde.sort_index()
up_sonde = up_sonde.sort_index()
down_sonde_resampled = down_sonde.resample("1min").interpolate()
up_sonde_resampled = up_sonde.resample("1min").interpolate()

down_const2 = down_constituents[[doc_col]].copy()
up_const2 = up_constituents[[doc_col]].copy()
down_sonde2 = down_sonde_resampled[[fdom_col]].copy()
up_sonde2 = up_sonde_resampled[[fdom_col]].copy()

# ensure datetime index
down_const2.index = pd.to_datetime(down_const2.index)
up_const2.index = pd.to_datetime(up_const2.index)
down_sonde2.index = pd.to_datetime(down_sonde2.index)
up_sonde2.index = pd.to_datetime(up_sonde2.index)
```

```python
# timestamp merge onto constituents index (keeps constituents timestamps only)
down_doc_fdom = down_const2.merge(down_sonde2, left_index=True, right_index=True, how="left")
up_doc_fdom = up_const2.merge(up_sonde2, left_index=True, right_index=True, how="left")
up_doc_fdom
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# Select only rows where both values exist
df = down_doc_fdom[['DOC (mg/L)', 'fDOM RFU']].dropna()

# X = fDOM, Y = DOC
X = df[['fDOM RFU']]
y = df['DOC (mg/L)']

# Fit regression
model = LinearRegression()
model.fit(X, y)

# Predictions
y_pred = model.predict(X)

# R²
r2 = r2_score(y, y_pred)

# Regression equation
slope = model.coef_[0]
intercept = model.intercept_

# Plot
plt.figure(figsize=(6, 5))

plt.scatter(df['fDOM RFU'], df['DOC (mg/L)'])

# Sort for a clean regression line
order = np.argsort(df['fDOM RFU'])
plt.plot(
    df['fDOM RFU'].iloc[order],
    y_pred[order]
)

plt.xlabel('fDOM (RFU)')
plt.ylabel('DOC (mg/L)')
plt.title('DOC vs fDOM')

plt.text(
    0.05, 0.95,
    f'DOC = {slope:.3f}·fDOM + {intercept:.3f}\n$R^2$ = {r2:.3f}',
    transform=plt.gca().transAxes,
    verticalalignment='top'
)

plt.tight_layout()
plt.show()

print(f'DOC = {slope:.4f} * fDOM + {intercept:.4f}')
print(f'R² = {r2:.4f}')
```

```python
from scipy.stats import linregress

result = linregress(df['fDOM RFU'], df['DOC (mg/L)'])

print(f"Slope = {result.slope:.4f}")
print(f"Intercept = {result.intercept:.4f}")
print(f"R² = {result.rvalue**2:.4f}")
print(f"r = {result.rvalue:.4f}")
print(f"p = {result.pvalue:.4e}")
```
