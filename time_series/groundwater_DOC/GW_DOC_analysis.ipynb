### Calculate hydraulic head (groundwater WSE - piezometer WSE)

```python
# For each piezometer, calculate VHG in each year
# VHG = groundwater WSE - piezometer WSE

vhg_results = {}
year_config = {
    2023: {
        "down": {
            "gw_left": "GW_down_left_2023",
            "gw_right": "GW_down_right_2023",
            "piezos": ["P1", "P2", "P3"],
            "piezo_suffix": "2023",
        },
    },
}
for year, banks in year_config.items():
    for bank, cfg in banks.items():
        gw_left = WSE_dict[cfg["gw_left"]][["Date_Time"] + cfg["piezos"]].rename(
            columns={p: f"{p}_gw_left" for p in cfg["piezos"]}
        )
        gw_right = WSE_dict[cfg["gw_right"]][["Date_Time"] + cfg["piezos"]].rename(
            columns={p: f"{p}_gw_right" for p in cfg["piezos"]}
        )

        for piezo in cfg["piezos"]:
            piezo_key = f"{piezo}C_{cfg['piezo_suffix'][-2:]}_WSE"
            piezo_df = WSE_dict[piezo_key][["Date_Time", "WSE"]].rename(columns={"WSE": f"{piezo}_WSE"})

            merged = (
                piezo_df.merge(gw_left[["Date_Time", f"{piezo}_gw_left"]], on="Date_Time", how="inner")
                        .merge(gw_right[["Date_Time", f"{piezo}_gw_right"]], on="Date_Time", how="inner")
            )

            merged[f"{piezo}_h_left"] = merged[f"{piezo}_gw_left"] - merged[f"{piezo}_WSE"]
            merged[f"{piezo}_h_right"] = merged[f"{piezo}_gw_right"] - merged[f"{piezo}_WSE"]
            merged[f"{piezo}_h_average"] = merged[[f"{piezo}_h_left", f"{piezo}_h_right"]].mean(axis=1)

            vhg_results[f"{piezo}_{bank}_{year}"] = merged[[
                "Date_Time",
                f"{piezo}_WSE",
                f"{piezo}_gw_left",
                f"{piezo}_gw_right",
                f"{piezo}_h_left",
                f"{piezo}_h_right",
                f"{piezo}_h_average"
            ]]

vhg_results["P1_down_2023"].head()
```

```python
# export data as csv files
output_dir = 'head_calculations'
for key, df in vhg_results.items():
    output_path = os.path.join(output_dir, f"{key}_hydraulic_head.csv")
    df.to_csv(output_path, index=False)
```

Calculating seasonal averages

```python
# only keep VHF averages for each piezometer: 
full_2023 = pd.DataFrame()
for key, df in vhg_results.items():
    if "2023" in key:
        piezo = key.split("_")[0]
        bank = key.split("_")[1]
        df_subset = df[["Date_Time", f"{piezo}_h_left", f"{piezo}_h_right", f"{piezo}_h_average"]]
        if full_2023.empty:
            full_2023 = df_subset
        else:
            full_2023 = full_2023.merge(df_subset, on="Date_Time", how="outer")

full_2023
```

```python
# separate 2023 data into seasons to calculate averages
spring_2023 = full_2023[(full_2023["Date_Time"] >= "2023-03-01") & (full_2023["Date_Time"] < "2023-06-19")]
summer_2023 = full_2023[(full_2023["Date_Time"] >= "2023-06-25") & (full_2023["Date_Time"] < "2023-09-21")]
# export as csv files
spring_2023.to_csv(os.path.join(output_dir, "spring_2023_hydraulic_head.csv"), index=False)
summer_2023.to_csv(os.path.join(output_dir, "summer_2023_hydraulic_head.csv"), index=False)
```

```python
# spring min, max and average
spring_stats = spring_2023.describe().loc[["min", "max", "mean"]]
spring_stats.to_csv(os.path.join(output_dir, "spring_2023_summary.csv"))
# summer min, max and average
summer_stats = summer_2023.describe().loc[["min", "max", "mean"]]
summer_stats.to_csv(os.path.join(output_dir, "summer_2023_summary.csv"))
```

Ploting hydraulic head time series 

```python
import matplotlib.dates as mdates

season_panels = [
    ("Spring", spring_2023),
    ("Summer", summer_2023),
]

piezo_cols = ["P1_h_average", "P2_h_average", "P3_h_average"]
piezo_names = [c.replace("_h_average", "") for c in piezo_cols]
colors = sns.color_palette("tab10", n_colors=len(piezo_cols))

fig, axes = plt.subplots(2, 1, figsize=(10, 5), sharex=False)

for ax, (season_name, season_df) in zip(axes, season_panels):
    season_df = season_df.copy()
    season_df["Date_Time"] = pd.to_datetime(season_df["Date_Time"])
    season_df = season_df.sort_values("Date_Time")

    for color, col, name in zip(colors, piezo_cols, piezo_names):
        if col in season_df.columns:
            ax.plot(
                season_df["Date_Time"],
                season_df[col],
                color=color,
                alpha=0.7,
                lw=1.2,
                label=name,
            )

    if "P123_h_average" in season_df.columns:
        ax.plot(
            season_df["Date_Time"],
            season_df["P123_h_average"],
            color="crimson",
            alpha=1.0,
            lw=1.5,
            label="average",
        )

    ax.set_title(f"{season_name} 2023")
    ax.set_ylabel("h")
    ax.grid(True, alpha=0.25)

    locator = mdates.AutoDateLocator()
    ax.xaxis.set_major_formatter(mdates.DateFormatter("%Y-%m-%d"))
    ax.xaxis.set_major_locator(locator)
    ax.tick_params(axis="x", labelbottom=True)

axes[1].set_xlabel("Date")
axes[0].legend(ncol=4, fontsize=9, frameon=False, loc="upper right")
plt.tight_layout()
plt.show()
```
