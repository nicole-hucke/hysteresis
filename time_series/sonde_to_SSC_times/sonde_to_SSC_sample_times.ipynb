```python
# testing hysteresis parameter calculation? 
import numpy as np
import pandas as pd
import os
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import MaxNLocator

sonde_downstream = pd.read_csv('../../data/sonde_data/sonde_down_full_record_smoothed.csv', parse_dates=['DateTime'], index_col='DateTime')
sonde_upstream = pd.read_csv('../../data/sonde_data/sonde_up_full_record_smoothed.csv', parse_dates=['DateTime'], index_col='DateTime')

spring_sample_times = pd.read_csv('spring_sample_times.csv', parse_dates=['Date'], index_col='Date')
summer_sample_times = pd.read_csv('summer_sample_times.csv', parse_dates=['Date'], index_col='Date')

# average rows with duplicate timestamps on sonde data (second-resolution data)
sonde_downstream = sonde_downstream.groupby(level=0).mean(numeric_only=True)
sonde_upstream = sonde_upstream.groupby(level=0).mean(numeric_only=True)

sonde_downstream_resampled = sonde_downstream.resample('1min').interpolate()
sonde_upstream_resampled = sonde_upstream.resample('1min').interpolate()
```

```python
# downstream sonde join
spring_sample_times = spring_sample_times.join(sonde_downstream_resampled, how="left")
summer_sample_times = summer_sample_times.join(sonde_downstream_resampled, how="left")
# change the column names to indicate that these are downstream sonde data (except for the first one)
spring_sample_times.columns = [spring_sample_times.columns[0], *[f"{col}_downstream" for col in spring_sample_times.columns[1:]]]
summer_sample_times.columns = [summer_sample_times.columns[0], *[f"{col}_downstream" for col in summer_sample_times.columns[1:]]]
```

```python
# upstream sonde join
spring_sample_times = spring_sample_times.join(sonde_upstream_resampled, how="left")
summer_sample_times = summer_sample_times.join(sonde_upstream_resampled, how="left")
# change the upstream sonde column names to indicate that these are upstream sonde data (except for the first three)
spring_sample_times.columns = [*spring_sample_times.columns[:3], *[f"{col}_upstream" for col in spring_sample_times.columns[3:]]]
summer_sample_times.columns = [*summer_sample_times.columns[:3], *[f"{col}_upstream" for col in summer_sample_times.columns[3:]]]
```

```python
# export to csv
spring_sample_times.to_csv('spring_sample_times_with_sonde.csv')
summer_sample_times.to_csv('summer_sample_times_with_sonde.csv')
```
